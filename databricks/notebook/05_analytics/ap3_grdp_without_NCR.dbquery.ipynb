```sql
%sql

SELECT
    fi.fact_indicator_id,
    r.region_id,
    r.region_code,
    r.region,
    y.year,
    fi.measurement AS grdp_per_capita,
    fe.total_obligation_amount
FROM pondolens.dev_gold.fact_regional_indicator fi

INNER JOIN pondolens.dev_gold.dim_year y
    ON fi.year_id = y.year_id

INNER JOIN pondolens.dev_gold.dim_regions r
    ON fi.region_id = r.region_id

LEFT JOIN (
    SELECT
        year_id,
        region_id,
        SUM(obligation_amount) AS total_obligation_amount
    FROM pondolens.dev_gold.fact_regional_expenditure
    GROUP BY year_id, region_id
) fe
    ON fi.year_id = fe.year_id
   AND fi.region_id = fe.region_id

WHERE fi.indicator = 'grdp' 
  AND r.region_code != 'NCR' 

ORDER BY y.year, r.region;
```
