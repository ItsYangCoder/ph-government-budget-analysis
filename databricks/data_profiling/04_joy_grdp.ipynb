## Silver Rule Proposal

For GRDP Silver, standardize one record per **geography + year + measure + price basis**.

- Keep **Current Prices** and **Constant 2018 Prices** separate.
- Keep **national** and **regional** records separate.
- Do not divide **Per-Capita GRDP** by population again.
- For **Total GRDP raw cells**, map verified headers, year columns, region rows, measures, units, and totals before creating analytical records.
- Preserve source lineage, revision information, and quality flags.
- Use GRDP in regional comparisons only when year and geography are compatible with the population and regional budget sources.
