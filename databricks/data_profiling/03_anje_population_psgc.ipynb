## Current Profiling Summary

- Out of the 6 bronze tables assigned, 2 will be used for silver: pondolens.bronze.psgc_2026_q2_national_summary_raw_cells and pondolens.bronze.population_projection_population
- pondolens.bronze.population_projection_population: 2052 rows. 100% null for source_page column.
- pondolens.bronze.psgc_2026_q2_national_summary_raw_cells: 288 rows. 100% null for source_page column. 67.01% nulls for parsed_numeric column. 
- Both tables contain all unique rows with 1 source_id and processing_revisions
- Bronze tables share lineage, source-location, parsed-value, dimension, and quality-flag fields.

## Silver Rule Proposal
- Only keep the 2 bronze tables population and PSGC national summary. Population contains needed population data while PSGC national summary contains the psgc codes at the regional level. Other PSGC tables only contain metadata on the PSGC table, PSGC codes up to the baranggay level, notes, and how to read the PSGC codes. These are unnecessary. 
- For PSGC national summary, the title of the columns of the table are at Row 10 from columns A to I. Column A contains CODES while Column C contains the REGION. Column B contains null values that can be discarded. 67.01% null parsed_numeric columns can be attributed to columns not having numeric values such as regions. The CODES in column A were not parsed in the parsed_numeric column.  
- For population table:
-- Regional totals: O and R at rows 28, 48, 68, 88, 108, 128, 148, 168, 188, 208, 228, 248, 268, 288, 308, 328, 348.
-- National control total: O8 and R8; do not add this to regional totals.
-- Both sexes: O and R, Male: P and S, Female Q and T
