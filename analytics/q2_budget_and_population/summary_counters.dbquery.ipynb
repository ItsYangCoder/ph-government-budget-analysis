```sql

CREATE OR REPLACE VIEW
pondolens.dev_gold.vw_ap2_summary_counters
AS

SELECT
    year,

    ROUND(MIN(per_resident), 0)
        AS lowest_per_resident,

    MIN_BY(region_name, per_resident)
        AS lowest_region,

    ROUND(
        TRY_DIVIDE(
            SUM(total_obligations),
            SUM(total_population)
        ),
        0
    ) AS national_per_resident,

    ROUND(MAX(per_resident), 0)
        AS highest_per_resident,

    MAX_BY(region_name, per_resident)
        AS highest_region,

    ROUND(
        TRY_DIVIDE(
            MAX(per_resident),
            MIN(per_resident)
        ),
        1
    ) AS highest_vs_lowest_x

FROM pondolens.dev_gold.vw_ap2_regional_alignment

GROUP BY year;
```
